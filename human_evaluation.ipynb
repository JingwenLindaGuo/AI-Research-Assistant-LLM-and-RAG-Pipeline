Load the models

```python
from openai import OpenAI
from dotenv import load_dotenv
import os
 
load_dotenv()
 
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

def embed_small(data):

    text_inputs = [d['text'] for d in data] if isinstance(data, list) else [data]

    response = client.embeddings.create(
        input=text_inputs,
        model="text-embedding-3-small"
    )
    return [{"values": item.embedding} for item in response.data]  

def embed_large(data):

    text_inputs = [d['text'] for d in data] if isinstance(data, list) else [data]

    response = client.embeddings.create(
        input=text_inputs,
        model="text-embedding-3-large"
    )
    return [{"values": item.embedding} for item in response.data]  
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

#cosine similarity
def cosine_sim(vec1, vec2):
    vec1 = np.array(vec1).reshape(1, -1)
    vec2 = np.array(vec2).reshape(1, -1)
    return cosine_similarity(vec1, vec2)[0][0]
```

Semantic similarity test

```python
test_sentences = [
    {"text": "The cat sat on the mat."},  #sentence 1
    {"text": "A feline rested on the rug."},  # similar to sentence 1
    {"text": "Quantum mechanics is a fundamental theory in physics."},  # not dimilar to sentence 1
]

small_embeddings = embed_small(test_sentences)
large_embeddings = embed_large(test_sentences)
small_vecs = [item["values"] for item in small_embeddings]
large_vecs = [item["values"] for item in large_embeddings]

small_sim_1_2 = cosine_sim(small_vecs[0], small_vecs[1])  # small 模型的句子1 vs 句子2
small_sim_1_3 = cosine_sim(small_vecs[0], small_vecs[2])  # small 模型的句子1 vs 句子3

large_sim_1_2 = cosine_sim(large_vecs[0], large_vecs[1])  # large 模型的句子1 vs 句子2
large_sim_1_3 = cosine_sim(large_vecs[0], large_vecs[2])  # large 模型的句子1 vs 句子3
```
