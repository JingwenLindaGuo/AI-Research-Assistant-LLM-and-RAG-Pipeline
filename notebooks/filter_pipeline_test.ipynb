## The hyphen error
- If a filter includes "-", the filter won't work.
- It applies to all filters.

```python
# Define filters using PaperFilter
filters = {'query': 'Artificial Intelligence',
    'author': 'ilya sutskever',
    "title": 'GPT-4o',
    "category": None,
    "abstract": None,
    "journal_ref": None,
    "doi": None,
    "exclude_words": None,
    'start_date': 201001010000,
    'end_date': 202512312359,
    'max_results': 10}

# Step 1: Fetch papers
papers = fetch_papers(filters)

# Step 2: Verify & Filter papers
filtered_papers = verify_results(papers, filters)

# Step 3: Chunk papers
paper_chunks = chunk_papers(filtered_papers)

# Print sample chunks
for i, chunk in enumerate(paper_chunks[:3]):
    print(f"\nChunk {i+1}:\n{chunk.page_content[:300]}...")

    
```
